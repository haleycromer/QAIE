# 04 Interpretation Notebook
Map latent embeddings to semantic, emotional, and material features.

```
import torch
from models.interpretation_network import InterpretationNetwork
import numpy as np

latent_dim = 128
interpreter = InterpretationNetwork(latent_dim=latent_dim, output_dim=15)

# Fake latent embedding from transformer
latent = torch.rand(1, latent_dim)
output = interpreter(latent)

features = ['Material', 'Emotion', 'Semantic'] + [f'Feature_{i}' for i in range(12)]
for f, v in zip(features, output.detach().numpy()[0]):
    print(f'{f}: {v:.3f}')
```
