# 07 Aircraft Detection Notebook
Classify synthetic aircraft types, trajectory, and material from multi-modal tensors.

```
import numpy as np
import matplotlib.pyplot as plt
from dataset.multimodal_synthetic import generate_multimodal_event
from preprocessing.tensor_fusion import multimodal_tensor

# Simulate 3 aircraft types
for i, freq in enumerate([440, 550, 660]):
    audio, vib, env = generate_multimodal_event()
    audio += 0.2*np.sin(2*np.pi*freq*np.linspace(0,1,len(audio)))
    tensor = multimodal_tensor(audio, vib, env)
    plt.figure(figsize=(10,3))
    plt.imshow(tensor[:,:,0].T, aspect='auto', cmap='plasma')
    plt.title(f'Aircraft Type {i+1}')
    plt.show()
```
