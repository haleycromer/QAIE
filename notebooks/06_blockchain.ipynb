# 06 Blockchain Notebook
Store AI observations in a ledger and verify distributed consensus.

```
from blockchain.distributed_verification import record_observation, verify_consensus
import torch

ledger_hashes = []
for i in range(3):
    obs = torch.rand(15)
    h = record_observation(obs)
    ledger_hashes.append(h)

print('Ledger hashes:', ledger_hashes)
consensus = verify_consensus()
print('Consensus hash:', consensus)
```
